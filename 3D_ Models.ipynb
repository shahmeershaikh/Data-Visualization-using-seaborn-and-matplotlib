```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# define data points
x_d = np.linspace(-1, 5, 10)
y_d = np.linspace(-1, 5, 10)

# compose
X, Y = np.meshgrid(x_d, y_d)
Z = 2 * X + Y + 5

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection = '3d')
ax.plot_surface(X, Y, Z, color = 'orchid', alpha = 0.5)
plt.show()
```

```python

# define data points
x = np.linspace(1, 10, 100)
y = np.linspace(1, 10, 100)

# compose
X, Y = np.meshgrid(x, y)

# 3rd dimension
Z = np.log(X**2 + Y**2)
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, cmap='cool')
plt.show()
```

```python
R = np.sqrt(X**2 + Y**2)
H = np.sin(R)

H = np.sin(np.sqrt(X**2 + Y**2))

# 3d
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, H, cmap='gnuplot')
plt.show()
```
