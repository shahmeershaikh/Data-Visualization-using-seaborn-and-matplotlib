```python
# Importing Libraries

import pandas as pd
import numpy as np
from wordcloud import WordCloud, STOPWORDS
import seaborn as sns
import matplotlib.pyplot as plt
import PIL.Image
```

```python
text_01 = """
Keep Healthy
Get A job
Own A business, Own A business, Own A business.
"""
print(text_01)
```

```python
print(STOPWORDS)
```

```python
plt.figure(figsize = (12, 6))
wcd_01 = WordCloud(stopwords = STOPWORDS, colormap = 'prism', background_color = '#81D8D0').generate(text_01)
plt.imshow(wcd_01)
plt.axis('off')
plt.show()
```

```python
poem = open('she_walks_in_beauty.txt', 'r').read()
print(poem)
```

```python
plt.figure(figsize = (15, 8))
wcd_02 = WordCloud(stopwords = STOPWORDS, colormap = 'prism', background_color = '#81D8D0').generate(poem)
plt.imshow(wcd_02)
plt.axis('off')
plt.show()
```

```python
pic_heart = np.array(PIL.Image.open('heart.png'))
```

```python
plt.figure(figsize = (5, 4))
wcd_02 = WordCloud(stopwords = STOPWORDS, colormap = 'prism', background_color = '#81D8D0',
                   mask = pic_heart, contour_color = 'blue', contour_width = 5).generate(poem)
plt.imshow(wcd_02)
plt.axis('off')
plt.show()
```
