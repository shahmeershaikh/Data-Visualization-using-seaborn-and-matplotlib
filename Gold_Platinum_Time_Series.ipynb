```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('annual_gold.csv')
df.head()
```

```python
df.dtypes
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
df.dtypes
```

```python
plt.figure(figsize = (14, 5))
sns.lineplot(data = df, x = 'Date', y = 'Gold', marker = 'o',
            markersize = 10, linestyle = '--', color = 'green')
plt.grid(axis = 'y')
plt.gcf().autofmt_xdate()
plt.show()
```

```python
plt.figure(figsize = (14, 5))
sns.lineplot(data = df, x = 'Date', y = 'Platinum', marker = 'X',
            markersize = 10, linestyle = '--', color = 'red')
plt.grid(axis = 'y')
plt.gcf().autofmt_xdate()
plt.show()
```

```python
plt.figure(figsize = (14, 5))
sns.lineplot(data = df, x = 'Date', y = 'Gold', marker = 'o',
            markersize = 10, linestyle = '--', color = 'green')
sns.lineplot(data = df, x = 'Date', y = 'Platinum', marker = 'X',
            markersize = 10, linestyle = '--', color = 'red')
plt.grid(axis = 'y')
plt.gcf().autofmt_xdate()
plt.show()
```
