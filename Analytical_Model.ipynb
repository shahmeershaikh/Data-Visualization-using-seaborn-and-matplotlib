```python
def xmax_demand(dates, a = 1, b = 1 ,c = np.pi * (3/4)):
    doy = dates.dayofyear
    demand = a + b * np.sin(2 * np.pi * doy/365 + c)
    return demand
```

```python
dates = pd.date_range('2026-01-01', '2026-12-31', freq = 'D')
demand = xmax_demand(dates)
```

```python
plt.figure(figsize = (10, 4))
plt.plot(dates, demand, color = 'teal', lw = 3)
plt.axhline(y = 0, color = 'red', linestyle = '--')
plt.axhline(y = 2, color = 'red', linestyle = '--')
plt.grid(True)
plt.show()
```

```python
n_steps = 200
x_steps = np.random.normal(size = n_steps)
y_steps = np.random.normal(size = n_steps)
```

```python
x_traj = np.cumsum(x_steps)
y_traj = np.cumsum(y_steps)
```

```python
plt.figure(figsize = (5, 4))
plt.plot(x_traj, y_traj, color = 'orange', lw = 3)
plt.show()
```

```python
  def lomax_pdf(r, alpha):
      return alpha * (1 + r) ** (-(alpha + 1))
```

```python
r = np.linspace(1, 10, 100)
alpha = 1.5
prob = lomax_pdf(r, alpha)
```

```python
plt.figure(figsize=(6, 3))
plt.plot(r, prob)

plt.axvline(6, color='grey', lw=1.2, ls='--')
plt.fill_between(r, prob, where=(r>=1) & (r<=6), color='gray', alpha=0.4)
plt.fill_between(r, prob, where=(r>=6) & (r<=10), color='gold', alpha=0.6)
plt.show()
```
