```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('annuaL_gold.csv')
df.head()
```

```python
df.dtypes
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Date_Ordinal'] = df['Date'].map(pd.Timestamp.toordinal)
df.head()
```

```python
df.dtypes
```

```python
plt.figure(figsize = (15, 4))
sns.lineplot(data = df, x = 'Date', y = 'Gold', marker = 'X',
            markersize = 10, color = 'goldenrod')
plt.grid(axis = 'y', lw = 0.7)
plt.show()
```

```python
z = np.polyfit(df['Date_Ordinal'], df['Gold'], 2)
p = np.poly1d(z)
```

```python
plt.figure(figsize = (15, 4))
sns.lineplot(data = df, x = 'Date', y = 'Gold', marker = 'X',
            color = 'goldenrod', markersize = 10, label = 'Price', lw = 2)
plt.plot(df['Date'], p(df['Date_Ordinal']), color = 'teal', lw = 4,
        label = 'Trend', alpha = 0.7)
plt.grid(axis = 'y', lw = 0.5)
plt.legend()
plt.gcf().autofmt_xdate()
plt.show()
```

```python
z = np.polyfit(df['Date_Ordinal'], df['Platinum'], 3)
p = np.poly1d(z)
```

```python
plt.figure(figsize = (15, 4))
sns.lineplot(data = df, x = 'Date', y = 'Platinum', marker = 'X',
            color = 'goldenrod', markersize = 10, label = 'Price', lw = 2)
plt.plot(df['Date'], p(df['Date_Ordinal']), color = 'green', lw = 4,
        label = 'Trend', alpha = 0.7)
plt.grid(axis = 'y', lw = 0.5)
plt.legend()
plt.gcf().autofmt_xdate()
plt.show()
```
