```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 1
t = np.linspace(0, 3, 100)
s_t = np.sin(t)
d_t = 1 / 10 * np.e ** t
```

```python
plt.figure(figsize=(9, 5))
plt.plot(t, s_t, color = 'green', linewidth = 2)
plt.plot(t, d_t, color = 'red', linewidth = 2)
plt.grid(axis = 'y', lw = 0.5)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.plot(t, s_t, color = 'green', label = 'Supply', linewidth = 2)
plt.plot(t, d_t, color = 'red', label = 'Demand', linewidth = 2)
plt.axvline(x = 0.1, color = 'grey', linestyle = '--')
plt.axvline(x = 2.14, color = 'brown', linestyle = '--')
plt.grid(axis = 'y', lw = 0.5)
plt.legend()
plt.show()
```
