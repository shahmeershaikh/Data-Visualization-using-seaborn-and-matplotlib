```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
attributes = ['Power','Speed','Strength',
              'Intelligence','Flight']
values = [90, 80, 75, 100, 90]
values = values + [90]
values
```

```python
angles = np.linspace(0, 2*np.pi, 5, endpoint = False).tolist()
angles = angles + [0]
angles
```

```python
plt.figure(figsize = (5, 4))
ax = plt.subplot(polar = True)
plt.xticks(angles[:-1], attributes)
ax.set_rlabel_position(0)
plt.yticks([20, 40, 60, 80, 100], color = 'purple', alpha = 0.7)
plt.ylim(0, 100)
ax.plot(angles, values, color = 'green', linewidth = 2)
ax.fill(angles, values, facecolor = 'skyblue', alpha = 0.3)
plt.show()
```

```python
values_flash = [95, 100, 90, 88, 59]
values_flash = values_flash + [95]
values_flash
```

```python
plt.figure(figsize=(5, 5))
ax = plt.subplot(polar=True)
# add attributes/features
plt.xticks(angles[:-1], attributes)
ax.set_rlabel_position(30)
plt.yticks([20, 40, 60, 80, 100], color='teal', alpha=0.8)
plt.ylim(0, 100)
# add wonder wonman's info
ax.plot(angles, values, label='wonder woman', color='pink', linewidth=3)
ax.fill(angles, values, facecolor='#cdb4db', alpha=0.5)
# add the flash
ax.plot(angles, values_flash, color='steelblue', linewidth=2, label='The Flash')
ax.fill(angles, values_flash, facecolor='skyblue', alpha=0.3)
plt.legend(loc='upper left')
plt.show()
```
